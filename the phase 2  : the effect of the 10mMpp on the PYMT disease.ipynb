```python
# Phase 2: the effect of 10mMpp treatment on PYMT disease 🧬

## 1. this is the analysis effect of 10mMpp treatment on the sample 1 affected
import pandas as pd

# Load the compressed RNA-Seq count data
df = pd.read_csv('GSE344396_Rai_et_al_rawcounts_allsamples.csv.gz', compression='gzip')

# Define columns for comparison (Replicate 1: Baseline vs. Day 7)
day0 =  'QGJ65Q_25_PYMT_LNP_10mMpp_Day0_Rep_1_Counts'

day7 =   'QGJ65Q_29_PYMT_LNP_10mMpp_Day7_Rep_1_Counts'



# 1. We calculate the difference between the first day and the seven day

df['Difference1'] = df[day7] - df[day0]

# 2 . we sort the table from largest to smallest

df_sorted = df.sort_values(by='Difference1', ascending=False)

# 3 . the genes that increased the most

top_3_up = df_sorted.head(10)

# 4 . the genes that decreased the most

top_3_down = df_sorted.tail(10)

# 5 . import all the genes in our table

important_genes = pd.concat([top_3_up, top_3_down])

# 6 . display all the information we need to know

important_genes[['Gene', day0, day7, 'Difference1']]
```

```python
 # Phase 2: the effect of 10mMpp treatment on PYMT disease 🧬

## 1. this is the analysis effect of 10mMpp treatment on the sample 4 affected
import pandas as pd

# Load the compressed RNA-Seq count data
df = pd.read_csv('GSE344396_Rai_et_al_rawcounts_allsamples.csv.gz', compression='gzip')

# Define columns for comparison (Replicate 1: Baseline vs. Day 7)
day0 =  'QGJ65Q_28_PYMT_LNP_10mMpp_Day0_Rep_4_Counts'

day7 =  'QGJ65Q_32_PYMT_LNP_10mMpp_Day7_Rep_4_Counts'


# 1. We calculate the difference between the first day and the seven day

df['Difference4'] = df[day7] - df[day0]

# 2 . we sort the table from largest to smallest

df_sorted = df.sort_values(by='Difference4', ascending=False)

# 3 . the genes that increased the most

top_3_up = df_sorted.head(10)

# 4 . the genes that decreased the most

top_3_down = df_sorted.tail(10)

# 5 . import all the genes in our table

important_genes = pd.concat([top_3_up, top_3_down])

# 6 . display all the information we need to know

important_genes[['Gene', day0, day7, 'Difference4']]
```

```python
# 1. Calculate the differences in disease in this file
df['Difference1'] = df['QGJ65Q_29_PYMT_LNP_10mMpp_Day7_Rep_1_Counts'] - df['QGJ65Q_25_PYMT_LNP_10mMpp_Day0_Rep_1_Counts']
df['Difference2'] = df['QGJ65Q_30_PYMT_LNP_10mMpp_Day7_Rep_2_Counts'] - df['QGJ65Q_26_PYMT_LNP_10mMpp_Day0_Rep_2_Counts']
df['Difference3'] = df['QGJ65Q_31_PYMT_LNP_10mMpp_Day7_Rep_3_Counts'] - df['QGJ65Q_27_PYMT_LNP_10mMpp_Day0_Rep_3_Counts']
df['Difference4'] = df['QGJ65Q_32_PYMT_LNP_10mMpp_Day7_Rep_4_Counts'] - df['QGJ65Q_28_PYMT_LNP_10mMpp_Day0_Rep_4_Counts']

# 2. Now calculate the average of the disease
df['Avg_Disease'] = (df['Difference1'] + df['Difference2'] + df['Difference3'] + df['Difference4']) / 4
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Calculate the average difference of the four mice
df['Average_Difference'] = (df['Difference1'] + df['Difference2'] + df['Difference3'] + df['Difference4']) / 4

# 2. Isolate the 10 genes that increased and became active (largest positive average)
top_up = df.sort_values(by='Average_Difference', ascending=False).head(10)

# 3. Isolate the 10 genes that decreased and were damaged (smallest negative average)
top_down = df.sort_values(by='Average_Difference', ascending=True).head(10)

# 4. Combine them
combined_genes = pd.concat([top_up, top_down])

# 5. Plot the graph
plt.figure(figsize=(10, 8))
colors = ['red' if x < 0 else 'green' for x in combined_genes['Average_Difference']]
plt.barh(combined_genes['Gene'], combined_genes['Average_Difference'], color=colors)

plt.xlabel('(Average Difference)')
plt.ylabel('(Genes)')
plt.title("Disease progression (upreglated with green and downregulated with red )")
plt.axvline(0, color='black', linewidth=1.2)
plt.show()
```
