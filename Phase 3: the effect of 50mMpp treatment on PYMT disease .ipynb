```python
import pandas as pd

# Load the compressed RNA-Seq count data
df = pd.read_csv('GSE344396_Rai_et_al_rawcounts_allsamples.csv.gz', compression='gzip')

# Define columns for comparison (Replicate 1: Baseline vs. Day 7)
day0 =  'QGJ65Q_33_PYMT_LNP_50mMpp_Day0_Rep_1_Counts'

day7 =  'QGJ65Q_37_PYMT_LNP_50mMpp_Day7_Rep_1_Counts'


# 1. We calculate the difference between the first day and the seven day

df['Difference1'] = df[day7] - df[day0]

# 2 . we sort the table from largest to smallest

df_sorted = df.sort_values(by='Difference1', ascending=False)

# 3 . the genes that increased the most

top_3_up = df_sorted.head(10)

# 4 . the genes that decreased the most

top_3_down = df_sorted.tail(10)

# 5 . import all the genes in our table

important_genes = pd.concat([top_3_up, top_3_down])

# 6 . display all the information we need to know

important_genes[['Gene', day0, day7, 'Difference1']]
```

```python
import pandas as pd

# Load the compressed RNA-Seq count data
df = pd.read_csv('GSE344396_Rai_et_al_rawcounts_allsamples.csv.gz', compression='gzip')

# 1. Calculate the differences in disease for this file so the computer can create and recognize the columns
df['Difference1'] = df['QGJ65Q_37_PYMT_LNP_50mMpp_Day7_Rep_1_Counts'] - df['QGJ65Q_33_PYMT_LNP_50mMpp_Day0_Rep_1_Counts']
df['Difference2'] = df['QGJ65Q_38_PYMT_LNP_50mMpp_Day7_Rep_2_Counts'] - df['QGJ65Q_34_PYMT_LNP_50mMpp_Day0_Rep_2_Counts']
df['Difference3'] = df['QGJ65Q_39_PYMT_LNP_50mMpp_Day7_Rep_3_Counts'] - df['QGJ65Q_35_PYMT_LNP_50mMpp_Day0_Rep_3_Counts']
df['Difference4'] = df['QGJ65Q_40_PYMT_LNP_50mMpp_Day7_Rep_4_Counts'] - df['QGJ65Q_36_PYMT_LNP_50mMpp_Day0_Rep_4_Counts']

# 2. Now calculate the average disease
df['Avg_Disease'] = (df['Difference1'] + df['Difference2'] + df['Difference3'] + df['Difference4']) / 4
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Calculate the average difference across the four mice
df['Average_Diff'] = (df['Difference1'] + df['Difference2'] + df['Difference3'] + df['Difference4']) / 4

# 2. Isolate the 10 genes that increased and were activated (largest positive average)
top_up = df.sort_values(by='Average_Diff', ascending=False).head(10)

# 3. Isolate the 10 genes that decreased and were destroyed (smallest negative average)
top_down = df.sort_values(by='Average_Diff', ascending=True).head(10)

# 4. Combine them
combined_genes = pd.concat([top_up, top_down])

# 5. Plot the graph
plt.figure(figsize=(10, 8))
colors = ['red' if x < 0 else 'green' for x in combined_genes['Average_Diff']]
plt.barh(combined_genes['Gene'], combined_genes['Average_Diff'], color=colors)

plt.xlabel('Average Difference')
plt.ylabel('Genes')
plt.title('Effect of 10mMpp treatment: Activated Genes (Green) and Destroyed Genes (Red)')
plt.axvline(0, color='black', linewidth=1.2)
plt.show()
```
